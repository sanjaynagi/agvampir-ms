```python
import pandas as pd

df_metadata = pd.read_csv("./config/metadata_ms.tsv", sep="\t")
df_metadata.head()
```

### compute association

```python
geno.shape
```

```python
df_metadata.shape
```

```python
gn = geno.compress(df_metadata.eval("country in ['Ghana', 'Kenya']"), axis=1)
gn
```

```python
df_metadata = df_metadata.query("country in ['Ghana', 'Kenya']")
```

```python
# convert to 0 and 1 
df_gn = pd.DataFrame(gn.to_n_alt())
```

```python
df_metadata.shape
```

```python
df_gn.T.shape
```

```python
df = pd.concat([df_metadata.reset_index(), df_gn.T], axis=1)
```

```python
df['country2'] = df['country'].map({'Ghana': 0, 'Kenya': 1})
```

```python
corrs = []
for i in range(95):
    corrs.append(df['country2'].corr(df[i]))
```

```python
pd.DataFrame(corrs).sort_values(0)
```
