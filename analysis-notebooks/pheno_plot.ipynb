```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("./results/config/metadata.qcpass.tsv", sep="\t").query("location == 'Siaya'")

df['status'] = df['sample_id'].apply(
    lambda x: 'Alive' if 'Alive' in str(x) else ('Dead' if 'Dead' in str(x) else 'Unknown')
)

# Count frequency
status_counts = df['status'].value_counts()

# Calculate percentages
total = status_counts.sum()
percent_alive = (status_counts.get('Alive', 0) / total) * 100
percent_dead = (status_counts.get('Dead', 0) / total) * 100

# Prepare data for plotly
plot_df = status_counts.reset_index()
plot_df.columns = ['status', 'count']

# Define color mapping
color_map = {
    'Alive': 'orange',
    'Dead': 'grey'
}

# Plot as pie chart with custom colors
fig = px.pie(
    plot_df,
    names='status',
    values='count',
    title='Proportion of Dead vs Alive',
    template='simple_white',
    width=400,
    color='status',
    color_discrete_map=color_map
)
fig.show()

# Print percentages
print(f"Percentage Alive: {percent_alive:.2f}%")
print(f"Percentage Dead: {percent_dead:.2f}%")
```
